```python
def is_russian_pangram(text):
    russian_alphabet = "абвгдежзийклмнопрстуфхцчшщъыьэюя"
    # Приводим текст к нижнему регистру и убираем пробелы и знаки препинания.
    text = ''.join(c for c in text.lower() if c in russian_alphabet)
    unique_letters = set(text)
    return len(unique_letters) == len(russian_alphabet)

text1 = "Съешь ещё этих мягких французских булок, да выпей чаю."
text2 = "Мама мыла раму"

print(f'"{text1}" - Панграмма: {is_russian_pangram(text1)}')
print(f'"{text2}" - Панграмма: {is_russian_pangram(text2)}')
```

```python
def find_two_sum(nums, target):
    num_map = {}

    # Перебираем массив.
    for index, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], index]
        num_map[num] = index
    return []

nums = [2, 7, 11, 15]
target = 9
result = find_two_sum(nums, target)
print(result)

nums = [3, 2, 4]
target = 6
result = find_two_sum(nums, target)
print(result)
```
